# 01 - Data Download

## This notebook imports the dataset and creates a clean copy/file path

```python
import pandas as pd
```

```python
from datasets import load_dataset

dataset = load_dataset("BoostedJonP/league_of_legends_match_data")

df = dataset["train"].to_pandas()

df.head()
```

```python
from pathlib import Path

directory = Path("../data/raw")
directory.mkdir(parents=True, exist_ok=True)

raw_path = directory / "league_matches.csv"
df.to_csv(raw_path, index=False)
```
